## What is Logistic Regression?

Imagine you're trying to predict whether a student will get admitted to college or not. The answer is either **YES (admitted)** or **NO (rejected)**. This is called a **binary classification** problem because there are only two possible outcomes.

**Logistic Regression** is a machine learning technique that helps us predict which of two categories something belongs to.

### Real-World Examples:
-  Will an email be spam or not spam?
-  Does a patient have a disease or not?
-  Will a student get admitted to college or not?
-  Will a customer buy a product or not?

---

## Linear Regression vs Logistic Regression

### Linear Regression
- Predicts **continuous numbers** (like temperature, price, age)
- Example: "What will be the house price?" → Answer: $250,000
- The prediction can be any number

### Logistic Regression
- Predicts **categories** (like yes/no, true/false, 0/1)
- Example: "Will the student be admitted?" → Answer: YES or NO
- The prediction is always one of two options

### Why Can't We Use Linear Regression for Classification?

Linear regression might predict values like -0.5 or 1.8, which don't make sense for a yes/no question! We need predictions between 0 and 1 that we can interpret as probabilities.

---

## Understanding Probabilities

Logistic regression gives us **probabilities** instead of direct yes/no answers.

For example:
- **Probability = 0.8** means 80% chance of being admitted
- **Probability = 0.3** means 30% chance of being admitted
- **Probability = 0.5** means 50-50 chance (right on the fence!)

By default, if the probability is:
- **Greater than 0.5** → Predict "Yes" (1)
- **Less than 0.5** → Predict "No" (0)

### What if the Probability is Exactly 0.5?

When the probability is exactly 0.5, it means our model is completely uncertain - it's like being right on the fence! In this case:

- Most implementations will classify it as "Yes" (1) by default
- But you can customize this threshold based on your needs
- Some applications may want to flag these borderline cases for human review

For example:
- In medical diagnosis, you might want to do additional tests
- In spam detection, you might want to let the email through
- In fraud detection, you might want to flag it for review

### Odds vs Probabilities

**Probability** and **odds** both describe how likely something is — but they say it in different ways. Logistic regression works with both behind the scenes, so it helps to know the difference.

| | **Probability** | **Odds** |
|---|-----------------|----------|
| **What it answers** | “What is the chance of success?” | “How many successes for each failure?” |
| **Range** | Between **0** and **1** (0% to 100%) | From **0** to **infinity** |
| **Example wording** | “80% chance of admission” | “4 admissions for every 1 rejection” |

**In plain words:**
- **Probability** is a slice of the whole pie (success out of *everything*).
- **Odds** compare **successes to failures only** — like a sports score: “4 to 1 in our favour.”

---

#### Example 1: College admission (probability → odds)

A model says a student has an **80% chance** of admission:

$$
{Probability} = 0.8
$$

**Convert to odds** (success ÷ failure):

$$
{odds} = \frac{0.8}{1 - 0.8} = \frac{0.8}{0.2} = 4
$$

**In plain words:** For every **1** student rejected, we expect about **4** admitted — odds of **4 to 1** in favour of admission.

---

#### Example 2: Same student (odds → probability)

Someone tells you the odds of admission are **4 to 1** (odds = 4). What is the probability?

$$
{probability} = \frac{\text{odds}}{1 + \text{odds}} = \frac{4}{1 + 4} = \frac{4}{5} = 0.8
$$

**In plain words:** 4 successes and 1 failure out of 5 “slots” → **80%** chance. Same answer as Example 1.

---

#### Quick reference: the two formulas

**Probability → odds**

$$
\text{odds} = \frac{p}{1 - p}
$$

**Odds → probability**

$$
p = \frac{\text{odds}}{1 + \text{odds}}
$$

*(Use the same \(p\) and odds in both directions — they are inverses of each other.)*

---

#### More examples (easy numbers)

| Situation | Probability \(p\) | Odds | How to read the odds |
|-----------|-------------------|------|----------------------|
| Very likely to be admitted | 0.90 | 9 | 9 admitted per 1 rejected |
| Coin flip / unsure | 0.50 | 1 | 1 admitted per 1 rejected (“even”) |
| Unlikely to be admitted | 0.20 | 0.25 | 1 admitted per 4 rejected (1 : 4) |
| Almost no chance | 0.10 | 0.11… | Roughly 1 success per 9 failures |

**Try it yourself — odds → probability:**  
Odds = **2** (2 successes per 1 failure):

$$
p = \frac{2}{1 + 2} = \frac{2}{3} \approx 0.67
$$

So a **67%** chance — about two-thirds likely.

---

#### Why this matters for logistic regression

The model often calculates **log-odds** first (a number that can be any value, positive or negative), then converts to **probability** so you get a number between 0 and 1. When you call `predict_proba()`, scikit-learn does that conversion for you — but the idea is the same: **odds and probability are two languages for the same belief**, and you can switch between them with the formulas above.

## Predicting a Binary Class

In our case we have two classes: `1=admitted` and `0=rejected`.

The logistic regression is still solving for $\hat{y}$. However, in our binary classification case, $\hat{y}$ will be the probability of $y$ being one of the classes.

$$
\hat{y} = P(y = 1)
$$

We'll still try to fit a "line" of best fit to this... except it won't be perfectly linear. We need to *guarantee* that the right-hand side of the regression equation will evaluate to a probability. (That is, some number between 0 and 1!)

# Fitting and making predictions with a logistic regression model in `scikit-learn`
---

We can follow the same steps to build a logistic regression model that we follow to build a linear regression model.

1. Define X & y
2. Instantiate the model
3. Fit the model
4. Generate predictions
5. Evaluate model

````python
# Step 1: Define X & y, and split into training & testing sets
X= admissions[['gpa']]
y= admissions['admit']

X_train, X_test, y_train, y_test = train_test_split(X,y,stratify = y, random_state = 50)
````

````python
# Step 2: Instantiate our model
logreg = LogisticRegression()

# Step 3: Fit our model
logreg.fit(X_train, y_train)
````

There are two methods in `sklearn` to be aware of when using logistic regression:
- `.predict()`
- `.predict_proba()`

````python
# Step 4 (part 1): Generate predicted values
````

````python
# Step 4 (part 2): Generate predicted probabilities

# Rounding the probabiloty to 3 decimal place
````

<details><summary>How would you interpret the predict_proba() output?</summary>
    
- This shows the probability of being rejected ($P(y=0)$) and the probability of being admitted ($P(y=1)$) for each observation in the testing data set.
- The first array, corresponds to the first testing observation.
    - The `.predict()` value for this observation is 0. This is because $P(y=0) > P(y=1)$.
- The second array, corresponds to the second testing observation.
    - The `.predict()` value for this observation is 0. This is because $P(y=0) > P(y=1)$.
</details>

````python
# Visualizing logistic regression probabilities
plt.figure()
plt.scatter(X_test,y_test, s=10)
plt.plot(X_test.sort_values('gpa'),
        logreg.predict_proba(X_test.sort_values('gpa'))[:,1],
        color='red', alpha=0.7, lw=3)

plt.xlabel('GPA')
plt.ylabel('Admit')
plt.title('Predicting Admissions from GPA');
````

# Interpreting the coefficients in a logistic regression model
---

````python
logreg.coef_
````

<details><summary> I want to interpret the coefficient $\hat{\beta}_1$ for my logistic regression model. How would I interpret this coefficient?</summary>
    
- Our model is that $\log\bigg(\frac{P(admit=1)}{1-P(admit=1)}\bigg) = \beta_0 + \beta_1\text{GPA}$.
- As GPA increases by 1, the log-odds of someone being admitted increases by 4.91.
- As GPA increases by 1, someone is $e^{4.92}$ times more likely to be admitted.
- As GPA increases by 1, someone is about 135 times more likely to be admitted to grad school.
</details>

> Hint: Use the [np.exp](https://docs.scipy.org/doc/numpy/reference/generated/numpy.exp.html) function.

````python
# Use np.exp() to exponentiate the coefficient.
np.exp(logreg.coef_)
````

## Many people prefer to think in probability instead of odds

Let's convert our result to a probability to make it easier to understand.

````python
# First of all, let's look at the intercept

logreg.intercept_
````

### Odds to Probabilities

Remember:

$$\ln \left({p\over 1-p}\right) = \beta_0 + \beta_1x$$

That means we'll get out the log odds if we compute the intercept plus the coefficient times a value for `GPA`.

**Compute the log odds of `admit` when `GPA=3`.**

### Converting to Real Probabilities
Let's say someone has a GPA of 3.5. To find their admission probability:

1. Calculate the "log odds": intercept + (coefficient × GPA)
2. Convert to regular odds: e^(log odds)
3. Convert to probability: odds / (1 + odds)


Don't worry about the math - the predict_proba() method does all this for you automatically!

You can confirm this is the same as the value you would get out of the `.predict_proba()` method of the `scikit-learn` object.

````python
# Compute predicted probability for GPA=3 using the predict_proba method.
logreg.predict_proba([[3.5]])
````

````python
# Increasing GPA by 0.5 (so that GPA=3.5)
logodds = logreg.intercept_ + logreg.coef_ * 3.5
odds = np.exp(logodds)
prob = odds/(1+odds)
print(prob)
````

**Interpretation:** A GPA score of 3.5 gives you a 43% chance of being admitted to college.

# What about Multiclass Classification?
---

Logistic regression is **only** for binary classification problems! However, there are two "hacks" that allow us to fit multiclass classification problems using the same/similar methodology. They are:

## Multinomial logistic regression
Normally, logistic regression is based off of the _Bernoulli distribution_, since our y-variable can be represented with a Bernoulli trial. We can instead base our y-variable off the _multinomial distribution_ and fit a model that way. It's more theoretically "pure" than OVR below, but less commonly used.

**How?**
```python
LogisticRegression(solver='...')
```
All solvers except `liblinear` will handle multiclass (the default is `lbfgs`).


## One-vs-Rest (OVR) logistic regression


A more common alternative is the _one-versus-rest_ technique. It works like this:

1. For category $k$, set $y = 1$ when the response is $k$ and $y = 0$ otherwise
1. Fit a logistic regression with this new $y$, and record the $\hat y$ for each observation
1. Repeat steps 1 and 2 for each category in our response variable
1. The true prediction is the $\hat y$ with the highest probability after all models are fit


**How?**

```python
sklearn.multiclass.OneVsRestClassifier()
```
You cannot do OVR in `LogisticRegression` in `scikit-learn`, but the library offers a `OneVsRestClassifier` classifier specifically for it: [`scikit-learn` documentation](https://scikit-learn.org/stable/modules/generated/sklearn.multiclass.OneVsRestClassifier.html#sklearn.multiclass.OneVsRestClassifier)

### 🤝 Shared Assumptions: Linear vs. Logistic Regression

| Shared Assumption | Description & Underlying Mechanism |
| :--- | :--- |
| **Structural Linearity** | Both models assume a straight-line relationship with the predictors ($X$). Linear regression fits this line directly to the **continuous outcome**, whereas logistic regression fits it to the **log-odds (logit)** of the outcome. |
| **Independence of Errors** | Both models require that all observations are sampled independently. The data must be free of clustering, temporal tracking (time-series), or repeated measurements on the same subject. |
| **No Multicollinearity** | Both models assume independent variables are not highly correlated with each other. Severe multicollinearity inflates standard errors and prevents the algorithms from calculating stable, isolated feature weights. |
| **Outlier Sensitivity** | Both models assume the absence of highly influential data points. Extreme outliers can disproportionately pull the linear regression line or skew the logistic decision boundary, degrading overall model generalization. |


# Deriving the Logistic Function from the Logit Function

How to invert the **Logit link function** (log-odds) to isolate probability ($p$), yielding the classic **Logistic (Sigmoid) function**.

Let $z$ represent the linear combination of inputs and weights:
$$z = \beta_0 + \beta_1x_1 + \beta_2x_2 + \dots + \beta_Kx_K$$


---

### Step 1: Start with the Logit Equation
* **The logistic regression model assumes that the log-odds of an observation y can be expressed as a linear function of the K input variables x**

The logit function defines the log-odds of a positive outcome as a linear function ($z$):

$$\ln\left(\frac{p}{1-p}\right) = z$$

Our objective is to solve this equation entirely for the probability parameter **$p$**.

### Step 2: Eliminate the Natural Logarithm ($\ln$)
Apply the exponential function (base $e$) to both sides of the equation to eliminate the natural log:

$$e^{\ln\left(\frac{p}{1-p}\right)} = e^z$$

Since $e^x$ and $\ln(x)$ are mathematical inverses, they cancel out on the left-hand side:

$$\frac{p}{1-p} = e^z$$

### Step 3: Clear the Fraction
Multiply both sides of the equation by the denominator $(1 - p)$ to remove the fraction:

$$p = e^z(1 - p)$$

Distribute $e^z$ across the terms inside the parentheses:

$$p = e^z - pe^z$$

### Step 4: Group the $p$ Terms Together
Move all terms containing $p$ to the left side of the equation by adding $pe^z$ to both sides:

$$p + pe^z = e^z$$

### Step 5: Factor Out $p$
Factor out the common variable $p$ from the left side of the equation:

$$p(1 + e^z) = e^z$$

### Step 6: Isolate the Probability ($p$)
Divide both sides by $(1 + e^z)$ to isolate $p$:

$$p = \frac{e^z}{1 + e^z}$$

### Step 7: Convert to Standard Logistic Form
To express this in standard machine learning notation, divide every term in both the numerator and the denominator by $e^z$:

$$p = \frac{\frac{e^z}{e^z}}{\frac{1}{e^z} + \frac{e^z}{e^z}}$$

Since $\frac{1}{e^z} = e^{-z}$ and $\frac{e^z}{e^z} = 1$, simplifying this expression gives us the final **Standard Logistic (Sigmoid) Function**:

$$p = \frac{1}{1 + e^{-z}}$$

---

### 🎯 Conclusion
The algebraic derivation shows that the logit and logistic functions are perfect mathematical mirrors of each other:
* **Logit function:** Maps a probability $p \in (0, 1)$ to an infinite linear space $z \in (-\infty, +\infty)$.
* **Logistic function:** Clamps an infinite linear space $z \in (-\infty, +\infty)$ back to a valid probability distribution $p \in (0, 1)$.
