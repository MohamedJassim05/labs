```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn import metrics
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import cross_val_score
%matplotlib inline
%config InlineBackend.figure_format = 'retina'
```

```python
# web location:
local_csv = './data/NHL_Data_GA.csv'
```

### 1. Load the NHL data

```python
# A:
df = pd.read_csv('./data/NHL_Data_GA.csv')
df
```

### 2. Perform any required data cleaning. Do some EDA.

### 3. Set up the `Rank` variable as your target. How many classes are there?

```python
# A:
y = df['Rank']
y.value_counts()
```

### 4. What is the baseline accuracy?

```python
# A:
y.value_counts(normalize=True).max()
```

### 5. Choose 4 features to be your predictor variables and set up your design matrix.

```python
# A:
df.corr(numeric_only =True)
```

```python
X = df[['GA','GA60','CA60','FA60']]
```

### 6. Fit a `KNeighborsClassifier` with 1 neighbor using the target and predictors.

```python
# A:
knn_model = KNeighborsClassifier(n_neighbors=1)
knn_model.fit(X,y)
```

### 7. Evaluate the accuracy of your model.
- Is it better than baseline?
- Is it legitimate?

```python
# A:
print(knn_model.score(X,y))
```

### 8. Create a 50-50 train-test-split of your target and predictors. Refit the KNN and assess the accuracy.

```python
# A:

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.5,random_state=42)
```

```python
knn_model.fit(X_train,y_train)
```

### 9. Evaluate the test accuracy of a KNN where K == number of rows in the training data.

```python
# A:

k = len(X_train)

knn = KNeighborsClassifier(n_neighbors=k)

knn.fit(X_train, y_train)

print('Test accuracy:', knn.score(X_test, y_test))
```

### 10. Fit the KNN at values of K from 1 to the number of rows in the training data.
- Store the test accuracy in a list.
- Plot the test accuracy vs. the number of neighbors.

```python
# A:
test_accuracies = []

for k in range(1, len(X_train) + 1):

    knn = KNeighborsClassifier(n_neighbors=k)

    knn.fit(X_train, y_train)

    accuracy = knn.score(X_test, y_test)

    test_accuracies.append(accuracy)

plt.figure(figsize=(10, 5))

plt.plot( range(1, len(X_train) + 1), test_accuracies)

plt.xlabel('Number of Neighbors (K)')
plt.ylabel('Test Accuracy')
plt.title('Test Accuracy vs Number of Neighbors')
plt.show()    
```

### 11. Fit KNN across different values of K and plot the mean cross-validated accuracy with 5 folds.

```python
# A:
mean_cv_accuracies = []

max_k = range(1,32,2)
for k in max_k:
    knn = KNeighborsClassifier(n_neighbors=k)
    mean_accuracy = cross_val_score(knn, X_train, y_train, cv=5).mean()
    mean_cv_accuracies.append(mean_accuracy)

plt.figure(figsize=(10, 5))
plt.plot(max_k, mean_cv_accuracies)
plt.xlabel('Number of Neighbors (K)')
plt.ylabel('Mean Cross-Validated Accuracy')
plt.title('Mean Cross-Validated Accuracy vs K')
plt.show()
```

### 12. Standardize the predictor matrix and cross-validate across the different K.
- Plot the standardized mean cross-validated accuracy against the unstandardized. Which is better?
- Why?

```python
# A:

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)


scaled_cv_accuracies = []

k_values = range(1, 32, 2)

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    mean_accuracy = cross_val_score( knn, X_train_scaled, y_train,cv=5).mean()
    scaled_cv_accuracies.append(mean_accuracy)


plt.figure(figsize=(10, 5))

plt.plot(k_values, mean_cv_accuracies, label='Unstandardized')
plt.plot(k_values, scaled_cv_accuracies, label='Standardized')

plt.xlabel('Number of Neighbors (K)')
plt.ylabel('Mean Cross-Validated Accuracy')
plt.title('Standardized vs Unstandardized KNN Accuracy')
plt.legend()
plt.show()
```

The standerized model , because it places all the predictors on similar scale, which leads each feature to contribute fairly.

## What was the best 'k'?

Using `GridSearch` & `Pipeline` find the best `k` for the KNN model

```python
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline
```

```python
knn_pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('knn', KNeighborsClassifier())
])

param_grid = {'knn__n_neighbors': range(1, 32, 2)}


grid_search = GridSearchCV(knn_pipeline,param_grid,cv=5,)

grid_search.fit(X_train, y_train)


print("Best K:", grid_search.best_params_['knn__n_neighbors'])
print("Best CV Accuracy:", grid_search.best_score_)
```
